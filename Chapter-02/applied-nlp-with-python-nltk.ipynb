## 2. Tokenization

Tokenization is the process of breaking text into smaller units called tokens.
NLTK provides `word_tokenize` (splits into words) and `sent_tokenize` (splits into sentences).

```python
input_text = "Carpe diem. Seize the day, boys. Make your lives extraordinary."

print(input_text)
```

```python
print(sent_tokenize(input_text))
```

## 3. Removing Punctuations and Stop Words

Stop words are common words (like 'the', 'is', 'at') that carry little meaningful information.
Removing them along with punctuation helps focus on the important words in the text.

```python
nltk.download('stopwords')
```

```python
from nltk.corpus import stopwords

input_text = """It was the best of times, it was the worst of times, it was the age of wisdom,
it was the age of foolishness, it was the epoch of belief, it was the epoch of incredulity,
it was the season of Light, it was the season of Darkness, it was the spring of hope,
it was the winter of despair."""

print(input_text)
```

```python
# Tokenize the text
tokens = word_tokenize(input_text)

# Remove punctuation
input_without_punctuation = [word for word in tokens if word.isalpha()]
print("After removing punctuation:")
print(input_without_punctuation)
```

```python
# Remove stop words
stop_words = set(stopwords.words('english'))
input_without_stop_words = [word for word in input_without_punctuation if word.lower() not in stop_words]
print("After removing stop words:")
print(input_without_stop_words)
```

## 4. Stemming

Stemming reduces words to their root/base form by removing suffixes.
We use the Snowball Stemmer which is an improvement over the original Porter Stemmer.

```python
from nltk.stem import snowball

stemmer = snowball.SnowballStemmer("english")
```

```python
input_text = "Learning does not make one learned: there is no substitute for experience and intuition."

# Tokenize
tokens = word_tokenize(input_text)

# Remove stop words
stop_words = set(stopwords.words('english'))
input_without_stop_words = [word for word in tokens if word.isalpha() and word.lower() not in stop_words]
print("Without stop words:", input_without_stop_words)

# Apply stemming
stemmed_words = [stemmer.stem(word) for word in input_without_stop_words]
print("After stemming:", stemmed_words)
```

## 5. Lemmatization

Lemmatization is similar to stemming but produces valid dictionary words (lemmas).
It uses vocabulary and morphological analysis to return the base form of a word.

```python
nltk.download('wordnet')
nltk.download('omw-1.4')
```

```python
from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()
```

```python
# Lemmatize the words (using input_without_stop_words from previous section)
lemmatized_words = [lemmatizer.lemmatize(word) for word in input_without_stop_words]
print("After lemmatization:", lemmatized_words)
```

## 9. Text Processing Pipeline (Twitter Sentiment)

Let's build a complete NLP pipeline to analyze sentiment in Twitter data.
We'll tokenize tweets, remove punctuation and stop words, and perform sentiment analysis.

```python
nltk.download("twitter_samples")
```

```python
from nltk.tokenize import TweetTokenizer
from nltk.corpus import twitter_samples
from collections import Counter
import matplotlib.pyplot as plt

# Get positive tweets
positive_tweets = twitter_samples.strings('positive_tweets.json')
negative_tweets = twitter_samples.strings('negative_tweets.json')

# Initialize tokenizer and sentiment analyzer
tweet_tokenizer = TweetTokenizer()
sia = SentimentIntensityAnalyzer()
stop_words = set(stopwords.words('english'))

def process_tweet(tweet):
    """Full NLP pipeline for a single tweet."""
    # Tokenize
    tokens = tweet_tokenizer.tokenize(tweet)
    # Remove punctuation
    tokens = [word for word in tokens if word.isalpha()]
    # Remove stop words
    tokens = [word for word in tokens if word.lower() not in stop_words]
    # Get sentiment
    sentiment = sia.polarity_scores(tweet)
    return tokens, sentiment

# Process sample tweets
print("Sample positive tweet:")
print(positive_tweets[0])
tokens, sentiment = process_tweet(positive_tweets[0])
print(f"Tokens: {tokens}")
print(f"Sentiment: {sentiment}")
```

```python
# Analyze sentiment distribution
sentiments = []
for tweet in positive_tweets[:100] + negative_tweets[:100]:
    _, sentiment = process_tweet(tweet)
    if sentiment['compound'] >= 0.05:
        sentiments.append('positive')
    elif sentiment['compound'] <= -0.05:
        sentiments.append('negative')
    else:
        sentiments.append('neutral')

# Plot results
sentiment_counts = Counter(sentiments)
plt.bar(sentiment_counts.keys(), sentiment_counts.values())
plt.title('Sentiment Distribution of Tweets')
plt.xlabel('Sentiment')
plt.ylabel('Count')
plt.show()
```

### TweetTokenizer vs word_tokenize Comparison

### Custom Amazon Go Chatbot

Let's build a custom chatbot that answers questions about Amazon Go stores.

```python
from nltk.chat.util import Chat, reflections

responses = (
    (r'What is Amazon Go\??',
     ('Amazon Go is a chain of convenience stores operated by Amazon. '
     'The stores use "just walk out" technology.',)),

    (r'How does Amazon Go work\??',
     ('Amazon Go uses computer vision, sensor fusion, and deep learning '
     'to detect when products are taken from or returned to shelves.',)),

    (r'Where (is|are) Amazon Go (store|stores|located)\??',
     ('Amazon Go stores are located in Seattle, Chicago, San Francisco, '
     'and New York City.',)),

    (r'What (can I|do you) (buy|sell|get) (at|in) Amazon Go\??',
     ('Amazon Go stores offer ready-to-eat breakfast, lunch, dinner, '
     'and snack options, as well as grocery essentials.',)),

    (r'Do I need (.*) to shop (at|in) Amazon Go\??',
     ('You need an Amazon account, a supported smartphone, and the '
     'Amazon Go app installed on your phone.',)),

    (r'(Hi|Hello|Hey)(.*)',
     ('Hello! I am the Amazon Go assistant. How can I help you?',
      'Hi there! Ask me anything about Amazon Go.',)),

    (r'(Bye|Goodbye|See you)(.*)',
     ('Goodbye! Have a great day.',
      'See you later! Enjoy your shopping experience.',)),

    (r'(.*)',
     ("I'm sorry, I don't understand. Try asking about Amazon Go stores, "
     "how they work, or what you can buy there.",)),
)
```

```python
def chat():
    """Start the Amazon Go chatbot."""
    print("Hi, I am the Amazon Go chatbot!")
    print("Ask me anything about Amazon Go. Type 'quit' to exit.\n")
    chatbot = Chat(responses, reflections)
    chatbot.converse()

# Uncomment the line below to start the chatbot
# chat()
```
